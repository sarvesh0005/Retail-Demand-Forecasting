```python
from pathlib import Path
import pandas as pd
```

```python
DATA_DIR = Path("../data/raw/m5")

calendar = pd.read_csv(DATA_DIR / "calendar.csv")
prices = pd.read_csv(DATA_DIR / "sell_prices.csv")
sales = pd.read_csv(DATA_DIR / "sales_train_validation.csv")
```

```python
datasets = {
    "Calendar": calendar,
    "Sell Prices": prices,
    "Sales": sales
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("=" * 60)
    print(f"Shape: {df.shape}")
    print(df.head())
    print("\n")
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("=" * 60)
    print(df.info())
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("=" * 60)
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print("=" * 70)
    print(name)
    print("=" * 70)
    print(df.describe(include="all"))
    print("\n")
```

```python
for name, df in datasets.items():
    print("=" * 70)
    print(name)
    print("=" * 70)
    print(df.nunique())
    print("\n")
```
